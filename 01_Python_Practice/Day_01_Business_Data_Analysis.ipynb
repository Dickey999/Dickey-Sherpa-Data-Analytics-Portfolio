```python
import pandas as pd

data = {
    "Customer": [
        "ABC Ltd", "XYZ Pvt Ltd", "Global Tech",
        "Prime Solutions", "ABC Ltd", "Future Corp",
        "XYZ Pvt Ltd", "Smart Systems"
    ],
    "Region": [
        "Kolkata", "Delhi", "Mumbai", "Kolkata",
        "Delhi", "Mumbai", "Kolkata", "Delhi"
    ],
    "Product": [
        "Laptop", "Desktop", "Laptop", "Desktop",
        "Laptop", "Laptop", "Desktop", "Laptop"
    ],
    "UnitsSold": [10, 5, 15, 8, 12, 20, 6, 9],
    "UnitPrice": [50000, 40000, 50000, 40000,
                  50000, 50000, 40000, 50000],
    "UnitCost": [38000, 30000, 38000, 30000,
                 38000, 38000, 30000, 38000]
}

df = pd.DataFrame(data)

df
```

```python
#Task 4: Calculate revenue.Create a new column called Revenue.
df['Revenue'] = df['UnitsSold'] * df['UnitPrice']
```

```python
#Task 5: Calculate total cost.Create a new column called TotalCost.
df['TotalCost']=df['UnitsSold'] * df['UnitCost']
```

```python
#Your task: Create a new column called Profit using Python.
df['Profit']=df['Revenue']-df['TotalCost']
```

```python
df[['Customer', 'Revenue', 'TotalCost', 'Profit']]
```

```python
#calculate total profit:
df['Profit'].sum()
```

```python
#Find the customer with the highest individual transaction profit:
df.loc[df['Profit'].idxmax()]
```

```python
#Next practical challenge — Customer profitability
#"Which customer generated the highest total profit across all their transactions?
df.groupby('Customer')['Profit'].sum()
```

```python
#Your manager wants to know the total revenue generated by each product, ranked from highest to lowest.
df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)
```
